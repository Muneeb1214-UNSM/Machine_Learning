```python
import plotly.express as px
```

```python
df =px.data.tips()
```

```python
df.head()
```

```python
fig = px.box(df,y="total_bill")
fig.show()
```

```python
fig = px.box(df,x="time",y="total_bill")
fig.show()
```

```python
fig = px.box(df,x="time",y="total_bill",points="all")
fig.show()
```

```python
fig = px.box(df,x="time",y="total_bill",color="smoker")
fig.show()
```
